# Task 3 : Data Detox - Data Cleaning Using Pandas

### 1.Loaded and Explored the Dataset

```python
import pandas as pd
df=pd.read_csv("netflix_titles.csv",encoding="latin1")
df.head()
```

### 2.Examined data types of all Columns

```python
df.info()
```

### 3.Identified missing values and duplicate records

```python
df.isnull().sum()
```

### 4. Handling Missing Values

```python
df['director']=df['director'].fillna("unknown")
df['cast']=df['cast'].fillna("unknown")
df['country']=df['country'].fillna("unknown")
df['date_added']=df['date_added'].fillna("unknown")
df['rating']=df['rating'].fillna("unknown")
df['duration']=df['duration'].fillna("unknown")
```

```python
df.head()
```

### 5.Fixed Text Inconsistencies

```python
df["type"]=df["type"].str.strip().str.title()
df["country"]=df["country"].str.strip()
```

### 6.Formatted columns correctly

```python
df["date_added"]=pd.to_datetime(df["date_added"],errors="coerce")
print(df["date_added"].head())
df["release_year"]=df["release_year"].astype(int)
```

### 7.Removed Duplicate Records

```python
df=df.drop_duplicates()
```

### 8. Removed Unnecessary Columns

```python
df=df.drop(columns=["Unnamed: 16","Unnamed: 17","Unnamed: 18","Unnamed: 19","Unnamed: 20","Unnamed: 21","Unnamed: 22"])
```

```python
df=df.drop(columns=["Unnamed: 12","Unnamed: 13","Unnamed: 14","Unnamed: 15","Unnamed: 23","Unnamed: 24","Unnamed: 25"])
```

```python
df.head()
```

### 9.Saved the Cleaned Dataset

```python
df.to_csv("cleaned_netflix_dataset.csv",index=False)
```
