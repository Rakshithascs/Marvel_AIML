```python
import pandas as pd
import random 
from faker import Faker
fake=Faker()
data=[]
for i in range(500):
    record={
        "Student ID":i+1,
        "Name":fake.name(),
        "Age":random.randint(18,25),
        "Gender":random.choice(["Male","Female"]),
        "Dept":random.choice(["cse","ece","ise","eee","mech","civil"]),
        "Attendance":random.randint(60,100),
        "IA_Marks":random.randint(40,100),
        "Final_Marks":random.randint(40,100),
        "Study_hours":random.randint(1,8)
    }
    data.append(record)
df=pd.DataFrame(data)
df.to_csv("student_performance_dataset.csv",index=False)
print("success")
```
